```python
data = pd.read_parquet('./data/green_tripdata_2025-01.parquet')
```

```python
from sklearn.linear_model import LinearRegression, LassoRegression, RidgeRegression
```

```python
data.head()
```

```python
data.lpep_pickup_datetime = pd.to_datetime(data.lpep_pickup_datetime)
data.lpep_dropoff_datetime = pd.to_datetime(data.lpep_dropoff_datetime)
```

```python
data['duration'] = data.lpep_dropoff_datetime - data.lpep_pickup_datetime
```

```python
data.duration = data.duration.apply(lambda td: td.total_seconds() / 60)
```

```python
data
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
sns.distplot(np.log1p(data['duration']))
```

```python
data.duration.describe()
```

```python
categorical = ['PULocationID', 'DOLocationID']
numerical = ['trip_distance']
```

```python
data[categorical] = data[categorical].astype(str)
```

```python
from sklearn.feature_extraction import DictVectorizer
dv = DictVectorizer()
```

```python
train_dicts = data[categorical + numerical].to_dict(orient='records')
```

```python
X_train = dv.fit_transform(train_dicts)
dv.feature_names_
```

```python
target = data.duration
y_train = target.values
y_train
```

```python
lr  = LinearRegression()
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_train)
sns.distplot(y_pred, label='prediction')
sns.distplot(y_train, label='actual')
plt.legend()
```
